```python
from pathlib import Path
import pandas as pd

main_dir = Path.cwd()
print(main_dir)
```

# Exploring GameTox

```python
PATH = "Existing_Datasets/GameTox/train.csv"
```

```python
df_gametox = pd.read_csv(main_dir / PATH )
```

```python
df_gametox.info()
```

```python
df_gametox = df_gametox[["message", "label"]]
```

```python
df_gametox.shape
```

```python
df_gametox = df_gametox.dropna()
df_gametox.shape
```

```python
df_gametox["label"] = df_gametox["label"].astype(int)
df_gametox.head()
```

```python
df_gametox["label"].value_counts()
```

```python
for label in sorted(df_gametox['label'].unique()):
    print(f"\n{'='*60}")
    print(f"LABEL {label} - {df_gametox[df_gametox['label'] == label].shape[0]} total samples")
    print('='*60)
    
    samples = df_gametox[df_gametox['label'] == label]['message'].head(10)
    
    for idx, text in enumerate(samples, 1):
        print(f"{idx:2d}. {text}")
```
